```python
#Inspecting the files position 
from pathlib import Path
import pandas as pd

# Locate the raw dataset folder
data_folder = Path("../data/raw")

# Find all CSV files
csv_files = sorted(data_folder.glob("*.csv"))

# Display the filenames
print("Number of CSV files found:", len(csv_files))

for file in csv_files:
    print(file.name)
```

```python

# Load each CSV file into a separate DataFrame
datasets = {}

for file in csv_files:
    df = pd.read_csv(file, low_memory=False)
    datasets[file.stem] = df

    print("\n" + "=" * 60)
    print("Dataset:", file.name)
    print("Number of rows:", df.shape[0])
    print("Number of columns:", df.shape[1])
    print("Column names:")
    print(df.columns.tolist())
```

```python
#finding null values, duplicates  and first 5 rows for each dataset
for name, df in datasets.items():
    print("\n" + "=" * 60)
    print("Dataset:", name)

    print("\nData types and non-null counts:")
    df.info()

    print("\nMissing values per column:")
    print(df.isnull().sum().sort_values(ascending=False))

    print("\nDuplicate rows:", df.duplicated().sum())

    print("\nFirst 5 records:")
    display(df.head())
```

```python
# Standardize URL formatting for comparison
main = datasets["ted_main"].copy()
transcripts = datasets["transcripts"].copy()


main["url_key"] = main["url"].astype(str).str.strip().str.rstrip("/")
transcripts["url_key"] = (
    transcripts["url"].astype(str).str.strip().str.rstrip("/")
)


print("Duplicate metadata URLs:", main["url_key"].duplicated().sum())
print("Duplicate transcript URLs:", transcripts["url_key"].duplicated().sum())

main_urls = set(main["url_key"])
transcript_urls = set(transcripts["url_key"])

matching_urls = main_urls.intersection(transcript_urls)

print("\nUnique metadata URLs:", len(main_urls))
print("Unique transcript URLs:", len(transcript_urls))
print("URLs found in both datasets:", len(matching_urls))
print("Metadata URLs without transcripts:", len(main_urls - transcript_urls))
```

```python
# Check duplicate URLs in transcripts
duplicate_transcripts = transcripts[
    transcripts["url_key"].duplicated(keep=False)
].sort_values("url_key")

print("Rows belonging to repeated URLs:", len(duplicate_transcripts))


duplicate_summary = (
    duplicate_transcripts.groupby("url_key")
    .agg(
        row_count=("transcript", "size"),
        unique_transcripts=("transcript", "nunique")
    )
    .reset_index()
)

display(duplicate_summary)
```

```python

quality_issues = pd.DataFrame([
    {
        "dataset": "ted_main",
        "issue": "Missing speaker_occupation",
        "count": 6,
        "planned_action": "Preserve the talks; fill missing occupations only if needed"
    },
    {
        "dataset": "transcripts",
        "issue": "Duplicate transcript URLs with identical text",
        "count": 3,
        "planned_action": "Keep one row per URL after confirming identical transcripts"
    },
    {
        "dataset": "combined data",
        "issue": "Metadata URLs without matching transcripts",
        "count": 86,
        "planned_action": "Keep metadata records; allow missing transcripts"
    }
])

display(quality_issues)
```

```python
#romoving the duplicate urls .
transcripts_clean = transcripts.copy()

transcripts_clean = transcripts_clean.drop_duplicates(
    subset="url_key",
    keep="first"
)


print("Original transcript rows:", len(transcripts))
print("Cleaned transcript rows:", len(transcripts_clean))

print(
    "Duplicate URLs remaining:",
    transcripts_clean["url_key"].duplicated().sum()
)
```

```python
#merging the two datasets on url_key to create a combined dataset

ted_combined = main.merge(
    transcripts_clean[["url_key", "transcript"]],
    on="url_key",
    how="left",
    validate="one_to_one"
)

print("Original metadata rows:", len(main))
print("Combined dataset rows:", len(ted_combined))
print("Missing transcripts:", ted_combined["transcript"].isna().sum())

display(
    ted_combined[
        ["title", "main_speaker", "url", "transcript"]
    ].head()
)
```

```python

# Summary statistics for numerical columns
display(ted_combined.describe())

# Convert duration from seconds to minutes for easier interpretation
duration_minutes = ted_combined["duration"] / 60

print("Average talk duration (minutes):", round(duration_minutes.mean(), 2))
print("Shortest talk (minutes):", round(duration_minutes.min(), 2))
print("Longest talk (minutes):", round(duration_minutes.max(), 2))

print("\nTalks with zero or negative duration:",
      (ted_combined["duration"] <= 0).sum())

print("Talks with zero or negative views:",
      (ted_combined["views"] <= 0).sum())
```

```python

# Count available transcripts
available = ted_combined["transcript"].notna().sum()
missing = ted_combined["transcript"].isna().sum()

print("Transcripts available:", available)
print("Transcripts missing:", missing)
print("Transcript coverage (%):",
      round(available / len(ted_combined) * 100, 2))

# Calculate word counts for available transcripts
word_counts = ted_combined["transcript"].dropna().str.split().str.len()

print("\nAverage transcript word count:", round(word_counts.mean(), 2))
print("Shortest transcript (words):", word_counts.min())
print("Longest transcript (words):", word_counts.max())
```

```python
# Calculate the number of words in each available transcript and display talks with fewer than 20 transcript words
ted_combined["transcript_word_count"] = (
    ted_combined["transcript"]
    .fillna("")
    .str.split()
    .str.len()
)

short_transcripts = ted_combined[
    ted_combined["transcript_word_count"].between(1, 19)
]

print("Talks with 1–19 transcript words:", len(short_transcripts))

display(
    short_transcripts[
        ["title", "main_speaker", "transcript_word_count", "transcript"]
    ].head(10)
)
```

```python

# Summary statistics for important numeric columns
display(
    ted_combined[["duration", "views", "comments", "languages"]]
    .describe()
)

# Check for potentially invalid values
print("Duration <= 0:", (ted_combined["duration"] <= 0).sum())
print("Views <= 0:", (ted_combined["views"] <= 0).sum())
print("Comments < 0:", (ted_combined["comments"] < 0).sum())
print("Languages < 0:", (ted_combined["languages"] < 0).sum())
```

 Exploratory Data Analysis — Findings

1. Dataset overview

- The metadata dataset contains 2,550 TED talks.
- The transcript dataset originally contains 2,467 rows.
- After removing duplicate transcript URLs, 2,464 unique transcript records remain.
- The datasets were merged using normalized URLs, preserving all 2,550 metadata records.

2. Transcript coverage

- Talks with transcripts: 2,464.
- Talks without transcripts: 86.
- Transcript coverage: 96.63%.
- Missing transcripts will be preserved as empty text during preprocessing so that these talks can still be recommended using their available metadata.

3. Data quality issues and decisions

- Missing speaker occupations: 6 records in the metadata dataset.
- Duplicate transcript URLs: 3 duplicate records identified and removed after confirming identical transcript text.
- Missing transcript matches: 86 metadata records have no matching transcript.
- Very short transcripts: 10 talks have between 1 and 19 transcript words. These records will be retained because some are performances and can still be represented using titles, descriptions, and tags.

4. Numerical analysis

- Numerical summary statistics were inspected using `describe()`.
- Talk duration was converted from seconds to minutes for easier interpretation.
- Zero or negative durations and views were checked. The observed counts are recorded in the notebook output.

5. Preprocessing decision

The original metadata will be preserved. Duplicate transcript records will be removed, metadata and transcripts will be combined using normalized URLs, and missing text fields will be handled during feature preparation. The cleaned dataset will retain talks even when transcripts are unavailable.

```python
# Create a separate copy for recommendation model preparation and fill missing text values with empty strings
ted_model_data = ted_combined.copy()

text_columns = ["title", "description", "tags", "transcript"]

for column in text_columns:
    ted_model_data[column] = ted_model_data[column].fillna("")

# Combine the available text fields into one column
ted_model_data["combined_text"] = (
    ted_model_data["title"] + " " +
    ted_model_data["description"] + " " +
    ted_model_data["tags"] + " " +
    ted_model_data["transcript"]
).str.strip()

# Check the prepared dataset
print("Dataset shape:", ted_model_data.shape)
print("Missing values in text columns:")
print(ted_model_data[text_columns].isna().sum())

display(
    ted_model_data[
        ["title", "description", "tags", "transcript", "combined_text"]
    ].head(3)
)
```

```python
# Check whether combined text was created correctly
print("Missing combined text values:",
      ted_model_data["combined_text"].isna().sum())

print("Empty combined text records:",
      (ted_model_data["combined_text"].str.len() == 0).sum())

print("\nSample combined text:")
print(ted_model_data["combined_text"].iloc[0][:500])
```

```python
# Save the prepared dataset to a CSV file in the dta/processed folder
from pathlib import Path


output_path = Path("../data/processed/ted_processed.csv")
ted_model_data.to_csv(output_path, index=False)

# Confirm the file was created
print("Dataset saved to:", output_path)
print("File exists:", output_path.exists())
```
